```python
def Planck_function(lambda_, T):
    h = 6.62607015e-27 # Planck constant in erg*s
    c = 29979245800 # Speed of light in cm/s
    k_B = 1.380649e-16 # Boltzmann constant in erg/K
    
    lambda_m = lambda_ # Don't Convert from Angstroms to cm
    exponent = (h * c) / (lambda_m * k_B * T)
    
    return (2 * h * c**2) / (lambda_m**5 * (np.exp(exponent) - 1))
```
