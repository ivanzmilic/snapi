```python
import numpy as np 
import matplotlib.pyplot as plt 
import pyana 
```

```python
falc51z = np.loadtxt("falc_51z.dat", unpack=True, skiprows=1)
mqs = pyana.fzread("mean_quiet_atmos.f0")["data"]
```

```python
# compare with pressure as the x coordinate
plt.figure(figsize=[8,5])
plt.plot(np.log10(falc51z[3]), falc51z[2], label='F')
plt.plot(np.log10(mqs[3,0,0]), mqs[2,0,0], label='MQS')
```
