```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.formatter.useoffset'] = False
```

```python
spectrum = np.loadtxt("falc_80z_500G_gris3.dat", unpack=True)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(121)
plt.plot(spectrum[1,:], label='I')
plt.subplot(122)
plt.plot(spectrum[4,:], label='V')
```

```python
spec1 = spectrum[:,:201]
spec2 = spectrum[:,201:602]
spec3 = spectrum[:,602:]
```

```python
spec1.shape
```

```python
# Now plot I, Q, V for all thee lines in a 3x3 grid: 

plt.figure(figsize=(13,8))
plt.subplot(3,3,1)
plt.plot(spec1[0,:]*1E8, spec1[1,:]/spec1[1,:].max(), label='I - K line')
plt.title('Stokes I for FALC atmosphere with 500G field')
plt.xlabel
plt.subplot(3,3,2)
plt.plot(spec1[0,:]*1E8, spec1[2,:]/spec1[1,:].max(), label='Q - K line')
plt.title('Stokes Q')
plt.subplot(3,3,3)
plt.plot(spec1[0,:]*1E8, spec1[4,:]/spec1[1,:].max(), label='V - K line')
plt.title('Stokes V')
plt.subplot(3,3,4)
plt.plot(spec2[0,:]*1E8, spec2[1,:]/spec2[1,:].max(), label='I - Ca II line')
plt.subplot(3,3,5)
plt.plot(spec2[0,:]*1E8, spec2[2,:]/spec2[1,:].max(), label='Q - Ca II line')
plt.subplot(3,3,6)
plt.plot(spec2[0,:]*1E8, spec2[4,:]/spec2[1,:].max(), label='V - Ca II line')
plt.subplot(3,3,7)
plt.plot(spec3[0,:]*1E8, spec3[1,:]/spec3[1,:].max(), label='I - Si I line')
plt.xlabel('Wavelength (Angstroms)')
plt.subplot(3,3,8)
plt.plot(spec3[0,:]*1E8, spec3[2,:]/spec3[1,:].max(), label='Q - Si I line')
plt.xlabel('Wavelength (Angstroms)')
plt.subplot(3,3,9)
plt.plot(spec3[0,:]*1E8, spec3[4,:]/spec3[1,:].max(), label='V - Si I line') 
plt.xlabel('Wavelength (Angstroms)')
plt.tight_layout()    
```
